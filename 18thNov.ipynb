```python
#Polymorphisam-->Poly means many, morphisam means logics
#concept of defining multiple logics or functionalities for performing a single operation is colled as polymorphism
#Static and Dynamic
#static: method overloading/constructor overloading
#dynamic: method overriding/constuctor over riding


```

```python
#method overloading: defining mulitple methods with same name and different name of parameters
#parameters
#under a single class
```

```python
class test:
  def m1(self):
    print("hi")
  def m1(self,a):
    print("how are you")
  def m1(self,a,b):
    print("good day")
t1=test()
t1.m1(2,3)
```

```python
#constuctor over loading
class test:
  def __init__(self):
    print("hi")
  def __init__(self,a):
    print("how are you")
  def __init__(self,a,b):
    print("good day")
t1=test(2,3)
```

```python
#method overriding:defining multiple methods with same name and same number of parameters one defined in one class and another defined in anthoer call
```

```python
class x:
  def m1(self):
    print("Hi Sai")
class y(x):
  def m1(self):
    print("Good day")
yobj=y()
yobj.m1()
```

```python
class x:
  def m1(self):
    print("Hi Sai")
class y(x):
  def m1(self):
    super().m1()
    print("Good day")
yobj=y()
yobj.m1()
```

```python
class x:
  def __init__(self):
    print("Hi Sai")
class y(x):
  def __init__(self):
    super().__init__()
    print("Good day")
yobj=y()
```
