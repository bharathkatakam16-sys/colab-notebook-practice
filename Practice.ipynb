```python
data={"Employee Name":["Raji","Ramani","Raghu","Siva","Neelu","Prame"],
      "empid":[102,304,598,654,542,543],
      "Age":[24,32,43,24,54,37],
      "Dept":["HR","finance","IT","Operator","Security","Data Analyst"]}
df=pd.DataFrame(data)
print(df)
```

```python
#print Data Analyst details
for i in df.Dept:
  if i == "Data Analyst":
    print(df.loc[df['Dept'] == 'Data Analyst','Employee Name'].to_string(index=False))
```

```python
#print Data Analyst Name, empid and age
for i in df.Dept:
  if i == "Data Analyst":
    print(df.loc[df['Dept'] == 'Data Analyst',['Employee Name','empid','Age']].to_string(index=False))
```
