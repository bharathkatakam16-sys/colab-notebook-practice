```python
i=1
while i<100:
  if i%5==0:
    print(i, end=" ")
  i+=1
```

for loop
-used to iterate over sequence or iterable
-often used when iterations are known
-automatically get next item
-supports break and continue

while loop
-used to repeat based on condition
-often used when iterations are unkown
-require manual condition updates
-supports break and continue
