```python
#Dictionary of the
import pandas as pd
import numpy as np
```

```python
data={"Name":['Sam','Suma','Suri','jam','ram','Krish'],
      "Age":[20,14,25,15,17,18]
      }
df=pd.DataFrame(data)
print(df)
```

```python
data={"Name":['Sam','Suma','Suri','jam','ram','Krish'],
      "Age":[20,14,25,15,17,18]
      }
df=pd.DataFrame(data,index=['r1','r2','r3','r4','r5','r6'])
print(df)
```

```python
#dt from list of dicts
data = [{'a':1,'b':2},{'a':5,'b':10,'c':20}]
df=pd.DataFrame(data)
print(df)
```

```python
data = [{'a':1,'b':2},{'a':5,'b':10,'c':20}]
df=pd.DataFrame(data, index=['1st','2nd'])
print(df)
```

```python
data = [{'a':1,'b':2},{'a':5,'b':10,'c':20}]
df1=pd.DataFrame(data, index=['1st','2nd'], columns=['a','b'])
print(df1)
```
