If and For loop


```python
#shopping discount check
amount=float(input("enter shopping amount: "))
membership=input("are you a member?(yes/no): ")
discount=0
if amount>=50000:
  if membership=="yes":
    discount=25
  else:
    discount=20
elif amount>=30000:
  if membership=="yes":
    discount=15
  else:
    discount=10
elif amount>=10000:
  if membership=="yes":
    discount=5
  else:
    discount=0
final_amount = amount-amount*discount/100
print("Discount: ", discount, '%')
print("Final amount: ", final_amount)
```

```python
n=5
for i in range(1,n+1):
  for j in range(1,i+1):
    if j==1 or j==i or i==n:
      print('*',end=" ")
  print()
```

```python
#pascal triangle
n=6
for i in range(n):
  num=1
  for j in range(n-i-1):
    print(" ", end="")
  for j in range(i+1):
    print(num, end=" ")
    num=num*(i-j)//(j+1)
  print()
```

```python
#floyds triangle
n=7
num=1
for i in range(1,n+1):
  for j in range(1,i+1):
    print(num, end=" ")
    num+=1
  print()
```

```python
#floyds triangle in reverse
n=7
num=1
for i in range(n,0,-1):
  for j in range(i):
    print(num, end=" ")
    num+=1
  print()
```
