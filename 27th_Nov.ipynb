```python
##basic plotting
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df=sns.load_dataset('tips')
sns.histplot(data=df,x='total_bill')
plt.title('histogram')
plt.show()
```

```python
df=sns.load_dataset('tips')
sns.histplot(data=df,x='total_bill',kde=True)
plt.title('histogram')
plt.show()
```

```python
df=sns.load_dataset('tips')
sns.histplot(data=df,x='total_bill')
plt.title('KDE plot')
plt.show()
```

```python
df=sns.load_dataset('tips')
sns.histplot(data=df,x='total_bill')
plt.title('EDE plot')
plt.show()
```

```python
sns.ecdfplot(data=df,x='total_bill')
plt.title('EDE plot')
plt.show()
```

```python
sns.displot(data=df,x='total_bill', kind='hist')
plt.title('dis plot')
plt.show()
```

```python
sns.scatterplot(data=df,x='total_bill',y='tip')
plt.title('Total bill vs tip')
plt.show()
```

```python
sns.scatterplot(data=df,x='total_bill',y='tip',hue='sex')
plt.title('Total_Bill')
plt.show()
```

```python
#line plot
sns.lineplot(data=df,x='total_bill', y='tip')
plt.title('Line Plot')
plt.show()
```

```python
#line plot with Grouping
sns.lineplot(data=df,x='total_bill', y='tip', hue='sex')
plt.title('Line Plot')
plt.show()
```

```python
sns.relplot(data=df,x='total_bill', y='tip', hue='sex',col='time', kind='scatter')
plt.title('Line Plot')
plt.show()
```

```python
#histogram - displays the fequency distribution of numerical data
#kdeplot - Kernal density estimation -- it display the smooth probability density distribution
#ecdfplot - empirical cumulative distribution function --- it shows the proportion of observations less than or equal to a particular value
#rugplot - it display small marks for individual observations
#displot - creates histogram, ecdf and kde plots
#scatterplot - relationship between 2 numerical values
#lineplot - visualize trends over continuous ordered data
#relational plot - a figure level function for scatter and line plots
```
