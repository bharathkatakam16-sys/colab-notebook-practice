```python
##basic plotting
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df=sns.load_dataset('tips')
sns.histplot(data=df,x='total_bill')
plt.title('histogram')
plt.show()
```

```python
df=sns.load_dataset('tips')
sns.histplot(data=df,x='total_bill',kde=True)
plt.title('histogram')
plt.show()
```

```python
df=sns.load_dataset('tips')
sns.histplot(data=df,x='total_bill')
plt.title('KDE plot')
plt.show()
```

```python
df=sns.load_dataset('tips')
sns.histplot(data=df,x='total_bill')
plt.title('EDE plot')
plt.show()
```

```python
sns.ecdfplot(data=df,x='total_bill')
plt.title('EDE plot')
plt.show()
```

```python
sns.displot(data=df,x='total_bill', kind='hist')
plt.title('dis plot')
plt.show()
```

```python
sns.scatterplot(data=df,x='total_bill',y='tip')
plt.title('Total bill vs tip')
plt.show()
```

```python
sns.scatterplot(data=df,x='total_bill',y='tip',hue='sex')
plt.title('Total_Bill')
plt.show()
```

```python
#line plot
sns.lineplot(data=df,x='total_bill', y='tip')
plt.title('Line Plot')
plt.show()
```

```python
#line plot with Grouping
sns.lineplot(data=df,x='total_bill', y='tip', hue='sex')
plt.title('Line Plot')
plt.show()
```

```python
sns.relplot(data=df,x='total_bill', y='tip', hue='sex',col='time', kind='scatter')
plt.title('Line Plot')
plt.show()
```

```python
#histogram - displays the fequency distribution of numerical data
#kdeplot - Kernal density estimation -- it display the smooth probability density distribution
#ecdfplot - empirical cumulative distribution function --- it shows the proportion of observations less than or equal to a particular value
#rugplot - it display small marks for individual observations
#displot - creates histogram, ecdf and kde plots
#scatterplot - relationship between 2 numerical values
#lineplot - visualize trends over continuous ordered data
#relational plot - a figure level function for scatter and line plots
#categorical plot
#barplot -- displays an aggregate statistic for each category
#countplot - display number of observations in each category

```

```python
sns.barplot(data=df,y='total_bill', x='day')
plt.title('bar Plot')
plt.show()
```

```python
sns.barplot(data=df,y='total_bill', x='day', hue='sex')
plt.title('bar Plot')
plt.show()
```

```python
sns.countplot(data=df,x='day')
plt.title('Count Plot')
plt.show()
```

```python
sns.countplot(data=df,x='day', hue='day')
plt.title('Count Plot')
plt.show()
```

```python
#difference between bar and count plot
#bar plot shows an aggregate value such as mean
#count plot show the number of records
```

```python
df=sns.load_dataset('tips')
print(df)
```

```python
#box plot -- isused to understand the spread of numerical data and identify outliers
#it display - minimum whisker
#           -1st quartile
#           -median
#           -3rd quartile
#           - maximum whisker
#           -potential outliers
```

```python
sns.boxenplot(data=df,x='day', y='total_bill')
plt.title('Box Plot')
plt.show()
#plt.title('Count Plot')
#plt.show()
```

```python
#violinplot -- it combines box plot with a distribution density shape
sns.violinplot(data=df,x='day', y='total_bill')
plt.title('Violin Plot')
plt.show()
```

```python
sns.violinplot(data=df,x='day', y='total_bill', hue='sex')
plt.title('Violin Plot')
plt.show()
```

```python
#strip plot -- it display individual values for categorical variables
sns.stripplot(data=df,x='day', y='total_bill')
plt.title('Strip Plot')
plt.show()
```

```python
sns.stripplot(data=df,x='day', y='total_bill', jitter=True)
plt.title('Strip Plot')
plt.show()
```

```python
#jitter - adding small random amount of variations to data points along an axis so that overlapping points become visible
```

```python
sns.stripplot(data=df,x='day', y='total_bill', jitter=0.1)
plt.title('Strip Plot')
plt.show()
```
