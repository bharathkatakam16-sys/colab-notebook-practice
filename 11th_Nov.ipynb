```python
from re import X
#function recursion/function innovations/recursive functions
def mymax(x,y,z):
  if x>y and x>z:
    return x
  elif y>x and y>z:
    return y
  else:
    return z
print(mymax(10,20,30))
```

```python
def mymax(x):
  if x==0:
    return 0
  elif x==1:
    return 1
  else:
    return x*mymax(x-1)
print(mymax(8))
```
