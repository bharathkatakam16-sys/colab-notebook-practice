```python
#factorial of a number
x=int(input("Enter a value for x: "))
fact=1
for i in range(1,x+1):
  fact=fact*i
print(fact)
```

```python
#count number of digits in a number
x=input("Enter a value for x:  ")
count=0
for digit in x:
  if digit.isdigit():
    count=count+1
print(count)
```

```python
#count number of vowels in a string
x=input("Enter a value for x:  ")
count=0
for ch in x:
  if ch in 'AEIOUaeiou':
    count=count+1
print(count)
```

```python
#largest num in the list
num=[10,45,78,3,278,32]
l=num[0]
for n in num:
  if n>l:
    l=n
print(l)
```

```python
#smallest num in the list
num=[10,45,78,3,278,32]
l=num[0]
for n in num:
  if n<l:
    l=n
print(l)
```

```python
#count positive num
num=[10,45,-78,3,-278,32]
count=0
for n in num:
  if n>0:
    count=count+1
print(count)
```

```python
x=[[1,2.3,4+5j],[True,"data",[1,2,3]]]
for i in x:
  print(i, type(i))
  for j in i:
    print(j, type(j))
```
