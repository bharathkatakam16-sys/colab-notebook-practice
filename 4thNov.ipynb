```python
#Modules
import copy
#hard copy
#shallow copy
#deep copy
x=[1,2,3]
a=x
b=copy.copy(x)
c=copy.deepcopy(x)
print(x)
print(a)
print(b)
print(c)
```

```python
#when i copy all the values gets copied in the same way but no difference
#how do you find the differences between hard, shallow and deep copies
```

```python
x=[1,2,3]
a=x
b=copy.copy(x)
c=copy.deepcopy(x)
x[1]=99
print(x)
print(a)
print(b)
print(c)
```

```python
#when can I see the change between shallow and deep
#when i have mutidimensional list probably i may have
```

```python
#to see the differences let us print the id's also -- storage address
print(x,id(x))
print(a,id(a))
print(b,id(b))
print(c,id(c))
```
