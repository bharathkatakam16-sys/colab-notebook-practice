```python
#categorical index
categories=pd.CategoricalIndex(['a','b','a','c'])
df=pd.DataFrame({'Col1':[20,30,50,60],'Col2':[30,50,60,70]},index=categories)
print(df)
print(df.index.dtype)
```

```python
#intervel index
intervel=pd.interval_range(start=0, end=4)
df=pd.DataFrame({'Col1':[20,30,50,60],'Col2':[30,50,60,70]},index=intervel)
print(df)
print(df.index.dtype)
```

```python
#multi index
arrays=[[1,1,2,2],['red','green','red','green']]
multiidx=pd.MultiIndex.from_arrays(arrays,names=('number','color'))
df=pd.DataFrame({'Col1':[20,30,50,60],'Col2':[30,50,60,70]},index=multiidx)
print(df)
print(df.index.dtype)
```

```python
#datetime index
datetime=pd.DatetimeIndex(['2022-01-01 11:00:00','2022-01-02 11:00:00'])
df=pd.DataFrame({'Col1':[20,30],'Col2':[30,50]},index=datetime)
print(df)
print(df.index.dtype)
```

```python
#timedelta index
timedelta_index=pd.timedelta_range(start='1 days',periods=5)
df=pd.DataFrame({'Col1':[20,30,50,60,70],'Col2':[45,65,67,89,90]},index=timedelta_index)
print(df)
print(df.index.dtype)
```
