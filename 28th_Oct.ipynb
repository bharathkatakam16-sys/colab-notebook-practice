```python
#if elif else
#consider 5 sub marks of student and cal his grade
s1=int(input("Enter a value for s1: "))
s2=int(input("Enter a value for s2: "))
s3=int(input("Enter a value for s3: "))
s4=int(input("Enter a value for s4: "))
s5=int(input("Enter a value for s5: "))
total=s1+s2+s3+s4+s5
print("Total: ", total)
per=total/5
print("Average: ", per)

if per>90:
  print("Grade: A+")
elif(per>=80 and per<90):
  print("Grade: A")
elif(per>=70 and per<80):
  print("Grade: A-")
elif(per>=60 and per<70):
  print("Grade: B")
elif(per>=50 and per<60):
  print("Grade: C")
else:
  print("FAIL")
```

```python
#0-100 -- 0/u
#100-200-- 10/u
#200 --- 25/u
x=int(input("Number of Units consumed: "))
if x<=100:
  amt=0
  print("Amount to be paid: ", amt)
elif(x>100 and x<=200):
  amt=(x-100)*10
  print("Amount to be paid: ", amt)
else:
  amt=1000+(x-200)*25
  print("Amount to be paid: ", amt)
```

```python
#nested if
x=int(input("Enter a value for x: "))
y=int(input("Enter a value for y: "))
z=int(input("Enter a value for z: "))
if x>y:
  if x>z:
    print("x is greater")
  else:
    print("z is greater")
else:
  if y>z:
    print("y is greater")
  else:
    print("z is greater")
```
