```python
#Inheritance -- we can access properties from one class to another either directly or indirectly
#is a relationship - 6types
#single inheritance
#Multilevel inheritance
#multiple inheritance
#cyclic inheritance
#hierarichal inheritance
#hybrid inheritance
```

```python
class x(object):
  def m1(self):
    print("method1")
class y(x):
  def m2(self):
    print("method2")
class z(y):
  def m3(self):
    print("method3")
yobj=y()
yobj.m1()
yobj.m2()
```

```python
#multilevel
class x(object):
  def m1(self):
    print("method1")
class y(x):
  def m2(self):
    print("method2")
class z(y):
  def m3(self):
    print("method3")
yobj=z()
yobj.m1()
yobj.m2()
yobj.m3()
```

```python
#Multiple
class x():
  def m1(self):
    print("method1")
class y():
  def m2(self):
    print("method2")
class z(x,y):
  def m3(self):
    print("method3")
yobj=z()
yobj.m1()
yobj.m2()
yobj.m3()
```

```python
#hirarchy
class x():
  def m1(self):
    print("method1")
class y(x):
  def m2(self):
    print("method2")
class z(y):
  def m3(self):
    print("method3")
yobj=y()
yobj.m1()
yobj.m2()

zobj=z()
zobj.m1()
zobj.m2()
```

```python
#cyclic
class x(z):
  def m1(self):
    print("method1")
class y(x):
  def m2(self):
    print("method2")
class z(y):
  def m3(self):
    print("method3")
yobj=z()
yobj.m1()
yobj.m2()
yobj.m3()
```

```python
#hybrid
class x():
  def m1(self):
    print("method1")
class y(x):
  def m2(self):
    print("method2")
class z(x):
  def m3(self):
    print("method3")
class a(y,z):
  def m4(self):
    print("method4")
yobj=a()
yobj.m1()
yobj.m2()
yobj.m3()
yobj.m4()
```
