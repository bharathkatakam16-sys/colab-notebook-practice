```python
#1
sal=int(input("Enter the sal: "))
if sal>=30000 and sal<60000:
  exp=float(input("Enter the exp:"))
if sal>=60000:
  rating = float(input("Enter the rating:"))
if sal<=30000:
  Bonus=sal*20/100
elif sal>=30000 and sal<60000 and exp>=5:
  Bonus = sal*15/100
elif sal>=30000 and sal<60000 and exp<=5:
  Bonus = sal*10/100
elif sal>=60000 and rating>=4.5:
  Bonus = sal*12/100
elif sal>=6000 and rating >=3.5 and rating<=4.49:
  Bonus = sal*8/100
else:
  Bonus="Not applicable"
print("Bonus is", Bonus)
```

```python
#2
year=input("enter the year: ")
yearsum=int(0)
for i in year:
  yearsum=yearsum+int(i)
print(yearsum)
if int(year)%4==0 and int(yearsum)%3==0:
  print("it's special leap year")
elif int(year)%4 == 0:
  print("it's leap year")
else:
  print("not a leap year")
```
